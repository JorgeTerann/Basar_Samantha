```python
# ============================================================
# TUPLAS EN PYTHON
# ============================================================


# ── 1. CREACIÓN DE TUPLAS ────────────────────────────────────

coordenada = (3.5, -1.2)
# Tupla de 2 elementos

dimensiones = (28, 28, 3)
# Alto, ancho y canales de una imagen RGB

unitaria = (42,)
# IMPORTANTE: la coma crea la tupla

no_tupla = (42)
# Esto NO es una tupla, es simplemente el entero 42


# ── 2. COMPROBAR EL TIPO ─────────────────────────────────────

print(type(unitaria))
# <class 'tuple'>

print(type(no_tupla))
# <class 'int'>


# ── 3. ACCESO POR ÍNDICE ─────────────────────────────────────

print(dimensiones[0])
# 28 → primer elemento

print(dimensiones[-1])
# 3 → último elemento


# ── 4. DESEMPAQUETADO (UNPACKING) ────────────────────────────

x, y = coordenada

print(x)
# 3.5

print(y)
# -1.2


alto, ancho, canales = dimensiones

print(alto)
# 28

print(ancho)
# 28

print(canales)
# 3


# ── 5. LAS TUPLAS SON INMUTABLES ──────────────────────────────

# Una vez creada una tupla, no podemos modificar sus elementos.

# dimensiones[0] = 64
# Esto produciría un TypeError.


# ── 6. USO REAL: SHAPE DE NUMPY ──────────────────────────────
# NumPy utiliza tuplas para representar las dimensiones
# de sus arrays.

import numpy as np

A = np.zeros((4, 5))

print(type(A.shape))
# <class 'tuple'>

print(A.shape)
# (4, 5)
```

```python
# ============================================================
# DICCIONARIOS EN PYTHON
# ============================================================


# ── 1. CREACIÓN ──────────────────────────────────────────────
# Un diccionario almacena información mediante pares:
# clave (key) → valor (value)

estudiante = {
    "nombre": "María",
    "edad": 22,
    "notas": [8.5, 9.0, 7.5],
    "activo": True
}


# ── 2. ACCESO A LOS VALORES ──────────────────────────────────

print(estudiante["nombre"])
# María → accede directamente a la clave "nombre"
# Si la clave no existe, produce un KeyError.


# ── 3. ACCESO SEGURO CON get() ───────────────────────────────

print(estudiante.get("carrera", "N/A"))
# N/A → si "carrera" no existe, devuelve "N/A"
# Evita que se produzca un KeyError.


# ── 4. MODIFICACIÓN Y ADICIÓN ────────────────────────────────

# Modificar un valor existente

estudiante["edad"] = 23


# Agregar una nueva clave

estudiante["semestre"] = 6


# Agregar otra clave

estudiante["carrera"] = "Ingeniería"


# Modificar una lista dentro del diccionario

estudiante["notas"].append(10.0)


# ── 5. VERIFICAR EXISTENCIA DE UNA CLAVE ─────────────────────

if "nombre" in estudiante:
    print("La clave existe")


# ── 6. ITERACIÓN ─────────────────────────────────────────────
# .items() permite obtener la clave y el valor
# al mismo tiempo.

for clave, valor in estudiante.items():
    print(f"{clave:12s} → {valor}")


# ── 7. DICT COMPREHENSION ────────────────────────────────────
# Permite crear diccionarios de forma compacta.
#
# Sintaxis:
# {clave: valor for elemento in iterable}

cuadrados = {
    x: x**2
    for x in range(1, 6)
}

print(cuadrados)
# {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}


# ── 8. MOSTRAR EL DICCIONARIO COMPLETO ───────────────────────

print(estudiante)
```

```python
# ============================================================
# FUNCIONES EN PYTHON
# ============================================================


# ── 1. FUNCIÓN BÁSICA ────────────────────────────────────────
# Una función se define con la palabra clave def.
# Puede recibir parámetros y devolver un resultado con return.

def saludar(nombre):
    return f"Hola, {nombre}!"


print(saludar("María"))
# Hola, María!


# ── 2. FUNCIÓN CON PARÁMETRO POR DEFECTO ─────────────────────
# Si no se proporciona un valor para "exponente",
# se utiliza el valor 2.

def potencia(base, exponente=2):
    """
    Eleva base a exponente.
    Por defecto calcula el cuadrado.
    """
    return base ** exponente


print(potencia(5))
# 25 → utiliza exponente=2

print(potencia(2, 10))
# 1024 → utiliza exponente=10


# ── 3. RETORNO MÚLTIPLE ──────────────────────────────────────
# Una función puede devolver varios valores.
# Python los empaqueta automáticamente en una tupla.

def min_max(lista):
    return min(lista), max(lista)


# Llamamos a la función

minimo, maximo = min_max([3, 1, 7, 2, 5])

# Los valores devueltos se desempaquetan
# en las variables minimo y maximo.

print(minimo, maximo)
# 1 7


# ── 4. LAMBDA — FUNCIÓN ANÓNIMA ──────────────────────────────
# Una lambda es una función pequeña y anónima.
#
# Sintaxis:
# lambda argumentos: expresión


doble = lambda x: x * 2

print(doble(8))
# 16


# Otro ejemplo de lambda

suma = lambda a, b: a + b

print(suma(3, 4))
# 7


# ── 5. LAMBDA COMO ARGUMENTO ─────────────────────────────────
# Las funciones como sorted() pueden recibir otra función
# mediante el parámetro key.


nums = [5, 2, 8, 1, 9, 3]

ordenados = sorted(
    nums,
    key=lambda x: -x
)

print(ordenados)
# [9, 8, 5, 3, 2, 1]
# Orden descendente


# ── 6. MAP ───────────────────────────────────────────────────
# map() aplica una función a cada elemento de un iterable.
#
# Sintaxis:
# map(función, iterable)


cuadrados = list(
    map(lambda x: x**2, nums)
)

print(cuadrados)
# [25, 4, 64, 1, 81, 9]


# ── 7. FILTER ────────────────────────────────────────────────
# filter() conserva únicamente los elementos
# que cumplen una condición.
#
# Sintaxis:
# filter(función, iterable)


mayores = list(
    filter(lambda x: x > 4, nums)
)

print(mayores)
# [5, 8, 9]
```

Ejercicio 3: Define una función 'resumen(lista)' que retorne un diccionario con:   longitud, suma, promedio, mínimo y máximo de la lista.

```python
def resumen(lista):
    longitud = len(lista)
    suma = sum(lista)
    promedio = suma / longitud
    minimo = min(lista)
    maximo = max(lista)

    return {
        "longitud": longitud,
        "suma": suma,
        "promedio": promedio,
        "mínimo": minimo,
        "máximo": maximo
    }


numeros = [10, 20, 30, 40, 50]

resultado = resumen(numeros)

print(resultado)
```

```python

import numpy as np

# Matriz 5x5 del 1 al 25
A = np.arange(1, 26).reshape(5, 5)
print("Matriz A original:")
print(A)

# a) Sub-matriz: 3 últimas filas y 3 primeras columnas
sub_matriz = A[-3:, :3]
print("\na) Sub-matriz (últimas 3 filas, primeras 3 columnas):")
print(sub_matriz)

# b) Segunda columna completa (índice 1)
columna_2 = A[:, 1]
print("\nb) Segunda columna completa:")
print(columna_2)

# c) Elementos de las esquinas
esquinas = A[::4, ::4]
print("\nc) Esquinas de la matriz:")
print(esquinas)
```

```python
#ejercio1
import numpy as np

# Matriz 5x4 con semilla 7
np.random.seed(7)
matriz = np.random.randint(1, 100, size=(5, 4))
print("Matriz generada:")
print(matriz)

# a) Media, máximo y mínimo por columna (axis=0)
media_col = matriz.mean(axis=0)
max_col = matriz.max(axis=0)
min_col = matriz.min(axis=0)
print("\na) Estadísticas por columna:")
print("Media:  ", media_col)
print("Máximo: ", max_col)
print("Mínimo: ", min_col)

# b) Suma acumulada de la primera fila
suma_acum = np.cumsum(matriz[0, :])
print("\nb) Suma acumulada de la primera fila:")
print(suma_acum)

# c) Normalización entre 0 y 1 por columna
normalizada = (matriz - min_col) / (max_col - min_col)
print("\nc) Matriz normalizada (0 a 1):")
print(normalizada.round(2))
```
